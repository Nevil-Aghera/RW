## What is Data Analysis?

Data analysis is the process of inspecting, cleaning, transforming, and modeling data with the goal of discovering useful information, drawing conclusions, and supporting decision-making. It includes exploratory data analysis (EDA) to understand distributions and relationships, descriptive statistics to summarize data, and inferential techniques to generalize findings. Effective data analysis combines domain knowledge, data-management skills, visualization, and clear communication of results.

## How to Plan a Data Science Project

A concise plan for a data science project typically includes:

- Define the objective: specify the business or research question and success metrics.
- Understand the data: list available data sources, what each field means, and any privacy/regulatory constraints.
- Data collection & ingestion: gather and centralize data from sources, ensure reproducibility.
- Data cleaning & preprocessing: handle missing values, outliers, inconsistent formats, and feature engineering.
- Exploratory Data Analysis (EDA): visualize distributions, correlations, and check assumptions.
- Modeling & validation: choose baseline models, evaluate with appropriate metrics and cross-validation, avoid data leakage.
- Deployment & monitoring: package model, deploy to production, and monitor performance and data drift.
- Communication: document methods, produce reports and visualizations for stakeholders.

## How to Frame a Machine Learning Problem

Framing an ML problem correctly is essential. Key steps:

- Determine the prediction target: classification, regression, ranking, or clustering.
- Define inputs and outputs: which features are available at prediction time.
- Choose evaluation metrics aligned with business goals (e.g., accuracy, precision/recall, AUC, RMSE).
- Decide the time horizon and data split strategy (temporal split if data is time-series).
- Identify constraints: latency, model size, interpretability, fairness, and privacy.
- Establish a baseline solution and iterate from there.

## Tensors — In-depth Explanation

A tensor is a generalization of scalars, vectors, and matrices to higher dimensions.

- Scalar: a 0-dimensional tensor (single number).
- Vector: a 1-dimensional tensor (array of numbers).
- Matrix: a 2-dimensional tensor (rows and columns).
- Higher-order tensors: arrays with 3 or more dimensions (e.g., a batch of images has shape [batch, height, width, channels]).

Important tensor concepts:

- Rank / order: the number of dimensions (axes) of the tensor.
- Shape: the size along each axis (e.g., (3, 4, 5)).
- dtype: the data type of elements (float32, int64, etc.).
- Broadcasting: rules that allow arithmetic between tensors of different shapes by implicitly expanding dimensions.

## 6. Handle missing data with:


```python
# 1. Simple Imputer (numerical: mean only).
from sklearn.impute import SimpleImputer

numeric_cols = df.select_dtypes(include=["int64", "float64"]).columns

mean_imputer = SimpleImputer(strategy="mean")

imputed_numeric = mean_imputer.fit_transform(df[numeric_cols])
df[numeric_cols] = pd.DataFrame(imputed_numeric, columns=numeric_cols)

print("Missing values after numerical mean imputation:")
print(df[numeric_cols].isna().sum())
```

```python
# 2. Simple Imputer (categorical: most frequent).

categorical_cols = df.select_dtypes(include=["object", "str"]).columns

freq_imputer = SimpleImputer(strategy="most_frequent")

imputed_categorical = freq_imputer.fit_transform(df[categorical_cols])
df[categorical_cols] = pd.DataFrame(imputed_categorical, columns=categorical_cols)

print("Missing values after categorical most-frequent imputation:")
print(df[categorical_cols].isna().sum())
```

```python
# 3. Most Frequent Category Imputation.

categorical_cols = df.select_dtypes(include=["object", "str"]).columns

for col in categorical_cols:
    most_frequent_value = df[col].mode()[0]
    df[col] = df[col].fillna(most_frequent_value)

print(df[categorical_cols].isna().sum())
```

```python
# 4. Missing Indicator + Random Sample Imputation.

cols_with_na = ['age', 'annual_income', 'credit_score']

indicator = MissingIndicator(features='all')
missing_flags = indicator.fit_transform(df[cols_with_na])

for i, col in enumerate(cols_with_na):
    df[col + '_was_missing'] = missing_flags[:, i]

np.random.seed(42)
for col in cols_with_na:
    missing_mask = df[col].isna()
    n_missing = missing_mask.sum()
    if n_missing > 0:
        random_samples = df[col].dropna().sample(n_missing, replace=True, random_state=42).values
        df.loc[missing_mask, col] = random_samples

print(df[[c + '_was_missing' for c in cols_with_na]].sum())
print(df[cols_with_na].isna().sum())
```

```python
# 5. KNN Imputer (multivariate).

numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

knn_imputer = KNNImputer(n_neighbors=5)
df[numeric_cols] = pd.DataFrame(knn_imputer.fit_transform(df[numeric_cols]), columns=numeric_cols)

print(df[numeric_cols].isna().sum())
```

```python
# 6. MICE Algorithm.


numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns

mice_imputer = IterativeImputer(max_iter=10, random_state=42)
df[numeric_cols] = pd.DataFrame(mice_imputer.fit_transform(df[numeric_cols]), columns=numeric_cols)

print(df[numeric_cols].isna().sum())
```

# 7. Detect and treat outliers using:


```python
# Z-score Method.


numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

col = 'annual_income'

z_scores = np.abs(stats.zscore(df[col]))
outliers_z = df[z_scores > 3]
df_z = df[z_scores <= 3]

print(outliers_z.shape)
print(df_z.shape)
```

```python
# IQR Method.

numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

col = 'annual_income'

Q1 = df[col].quantile(0.25)
Q3 = df[col].quantile(0.75)
IQR = Q3 - Q1
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers_iqr = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
df_iqr = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

print(outliers_iqr.shape)
print(df_iqr.shape)
```

```python
# Percentile Method.

numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

col = 'annual_income'

lower_percentile = df[col].quantile(0.01)
upper_percentile = df[col].quantile(0.99)

df_pct = df.copy()
df_pct[col] = np.clip(df_pct[col], lower_percentile, upper_percentile)

print(lower_percentile, upper_percentile)
print(df_pct[col].min(), df_pct[col].max())
```

```python
# Winsorization Technique.

numeric_cols = df.select_dtypes(include=['int64', 'float64']).columns
df[numeric_cols] = df[numeric_cols].fillna(df[numeric_cols].mean())

col = 'annual_income'

df_win = df.copy()
df_win[col] = winsorize(df_win[col], limits=[0.01, 0.01])

print(df_win[col].min(), df_win[col].max())
```

# 12. Apply transformations:


```python
# FunctionTransformer log transform, reciprocal, square root.

df['annual_income'] = df['annual_income'].fillna(df['annual_income'].mean())

log_transformer = FunctionTransformer(np.log1p)
df['annual_income_log'] = log_transformer.fit_transform(df[['annual_income']])

reciprocal_transformer = FunctionTransformer(lambda x: 1 / (x + 1))
df['annual_income_reciprocal'] = reciprocal_transformer.fit_transform(df[['annual_income']])

sqrt_transformer = FunctionTransformer(np.sqrt)
df['annual_income_sqrt'] = sqrt_transformer.fit_transform(df[['annual_income']])

print(df[['annual_income', 'annual_income_log', 'annual_income_reciprocal', 'annual_income_sqrt']].head())
```

```python
# PowerTransformer Box-Cox and Yeo-Johnson.

boxcox = PowerTransformer(method='box-cox')
df['annual_income_boxcox'] = boxcox.fit_transform(df[['annual_income']])

yeojohnson = PowerTransformer(method='yeo-johnson')
df['annual_income_yeojohnson'] = yeojohnson.fit_transform(df[['annual_income']])

print(df[['annual_income', 'annual_income_boxcox', 'annual_income_yeojohnson']].head())
print('Box-Cox lambda:', boxcox.lambdas_)
print('Yeo-Johnson lambda:', yeojohnson.lambdas_)
```

```python
# Column Transformer apply different preprocessing steps to different columns.

numeric_features = ['age', 'annual_income', 'loan_amount', 'credit_score']
nominal_features = ['region', 'loan_purpose']
ordinal_features = ['education_level']

education_order = ['Primary', 'Secondary', 'Graduate', 'Post-Graduate']

numeric_pipeline = Pipeline([
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler())
])

ordinal_pipeline = Pipeline([
    ('encoder', OrdinalEncoder(categories=[education_order]))
])

preprocessor = ColumnTransformer(transformers=[
    ('num', numeric_pipeline, numeric_features),
    ('nom', OneHotEncoder(handle_unknown='ignore'), nominal_features),
    ('ord', ordinal_pipeline, ordinal_features)
])

transformed = preprocessor.fit_transform(df)
print('Transformed shape:', transformed.shape)
print(preprocessor.get_feature_names_out()[:15])
```

# 15. Write a report summarizing:


### Missing Value Strategies Used and Their Effectiveness

Five columns contained missing values: `age` (12), `gender` (10), `employment_type` (8), `annual_income` (7), and `credit_score` (6).

- **Numeric columns** → **KNN Imputer (k=5)**: chosen over simple mean/median because it estimates missing values from similar customers across multiple correlated features (e.g. income and credit score together), preserving relationships a flat average would destroy.
- **Categorical columns** → **Most Frequent (mode) imputation**: appropriate since each column had a dominant category and missingness was low (~3%).

**Effectiveness:** all 43 originally missing cells were resolved, leaving 0 missing values in the final dataset. Because missingness was under 5% per column, no rows/columns needed to be dropped — Complete Case Analysis was evaluated but rejected, since it would have discarded 31 rows (~10% of the data).

### Outlier Handling Results

Z-score (|z| > 3) and IQR (1.5×IQR) methods were applied to `annual_income`, `loan_amount`, `credit_score`, and `spending_ratio`. **Both methods detected 0 outliers** across all four columns — the dataset appears to be synthetically generated within realistic, tightly-bounded ranges.

As a safeguard, **Winsorization (5%/5%)** was still applied to compress extreme tail values slightly (e.g., `annual_income` min moved from ~226K to ~394K), protecting the pipeline against genuine outliers in future/unseen data batches, without materially changing the current distribution.

### Encoding Methods Applied

| Column | Method | Reason |
|---|---|---|
| `education_level` | Ordinal Encoding | Natural rank order: Primary < Secondary < Graduate < Post-Graduate |
| `region` | One-Hot Encoding | Nominal, no inherent order (4 categories) |
| `loan_purpose` | One-Hot Encoding | Nominal, no inherent order (5 categories) |
| `gender` | One-Hot Encoding | Nominal (3 categories) — no binary column existed for Label Encoding |
| `employment_type` | One-Hot Encoding | Nominal (3 categories) |
| `customer_id` | Mixed-variable split | Alphanumeric ID split into numeric part + category prefix |

Label Encoding was considered for binary features, but no categorical column in this dataset is strictly binary. `default_flag` — the only true binary variable — is already numeric and needed no encoding.


### Scaling / Transformations Applied and Why

- **Standardization (Z-score scaling)**: applied to all continuous numeric features in the scaled export. Centers each feature at mean 0 / std 1 — important for distance-based and gradient-based models (logistic regression, SVM, KNN, neural nets), and prevents a large-magnitude column like `annual_income` (in the millions) from dominating smaller-scale features.
- **Min-Max, MaxAbs, and Robust scaling** were also demonstrated; Robust scaling is recommended if future data batches contain genuine outliers, since it uses median/IQR instead of mean/std.
- **Log transform (log1p) and Box-Cox / Yeo-Johnson power transforms** were tested on `annual_income` to reduce right-skew — Box-Cox and Yeo-Johnson gave near-identical results (λ ≈ 0.77) since income is strictly positive.
- A **ColumnTransformer** formalized the pipeline: numeric columns get imputation + scaling, nominal columns get one-hot encoding, and `education_level` gets ordinal encoding — all combined into one reusable, leakage-safe transform.

### Newly Engineered Features and Their Usefulness

| Feature | Formula | Usefulness |
|---|---|---|
| `debt_to_income_ratio` | `loan_amount / annual_income` | Core credit-risk indicator; higher ratios typically signal greater default risk |
| `avg_monthly_transactions` | `transaction_count / 12` | Proxy for financial activity / engagement level |
| `spending_to_income_ratio` | `spending_ratio / 100` | Normalized spending behavior relative to income |
| `join_year` / `join_month` / `join_day` / `join_weekday` | extracted from `join_date` | Enables tenure and seasonality analysis |
| `education_level_encoded` | ordinal mapping | Numeric representation preserving education rank |

`debt_to_income_ratio` in particular is a standard, high-importance predictor in real-world credit risk models.

### Final Dataset Shape and Readiness for ML Modeling

- **Shape:** 300 rows × 34 columns (up from the original 300 × 15), with 0 missing values remaining.
- All categorical variables are now fully numeric-encoded (ordinal or one-hot); only `customer_id` and `join_date` remain non-numeric, and both are identifiers that should be excluded from the feature matrix, not modeling features.
- Two versions were exported:
  - `cleaned_credit_risk_dataset.csv` — unscaled, suited to tree-based models (Random Forest, XGBoost, etc.)
  - `cleaned_scaled_credit_risk_dataset.csv` — standardized, suited to distance/gradient-based models
- **Class imbalance note:** `default_flag` is 71% positive class. This should be addressed at modeling time (class weighting, SMOTE/resampling, or stratified train-test splits) rather than in preprocessing.



**Conclusion:** the dataset is clean, fully numeric, free of missing values, and enriched with domain-relevant engineered features — ready to be split into train/test sets and passed into a classification model to predict `default_flag`.
