```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

```

```python
student_data = pd.read_csv('D:\RW\Mathematic\project-2\Data\student_exam_dataset.csv')

student_data.head()
```

```python

# Question 1 - Understanding the Basics

print("What is Probability?")
print("Probability is the chance of occurrence of an event.")

print("\nKey Probability Terminology")
print("1. Experiment")
print("2. Outcome")
print("3. Event")
print("4. Sample Space")
print("5. Probability")

print("\nProbability Event Examples")
print("1. Student passes exam")
print("2. Student attendance above 80%")
print("3. Student participates in group discussion")
```

```python

# Question 2 - Types of Events

# Empirical Probability
empirical_probability = (student_data['final_exam_pass'] == 'Pass').mean()
print("Empirical Probability of Passing =", empirical_probability)

# Theoretical Probability
theoretical_probability = 1/2
print("Theoretical Probability =", theoretical_probability)
```

```python

# Question 3 - Random Variable & Probability Distribution

from collections import Counter
from math import comb

# Random variable:
# Number of students passing among 3 randomly selected students

p = (student_data['final_exam_pass'] == 'Pass').mean()

distribution = []

for x in range(4):
    probability = comb(3, x) * (p**x) * ((1-p)**(3-x))
    distribution.append([x, probability])

distribution_df = pd.DataFrame(distribution, columns=['X', 'P(X)'])

print(distribution_df)

mean = 3 * p
variance = 3 * p * (1-p)

print("\nMean =", mean)
print("Variance =", variance)
```

```python

# Question 4 - Venn Diagram in Probability

study_more_10 = student_data[student_data['study_hours'] > 10]
attendance_more_80 = student_data[student_data['attendance'] > 80]

overlap = student_data[
    (student_data['study_hours'] > 10) &
    (student_data['attendance'] > 80)
]

print("Students studying more than 10 hours =", len(study_more_10))
print("Students attendance above 80 =", len(attendance_more_80))
print("Overlap students =", len(overlap))

from matplotlib_venn import venn2

venn2(subsets = (
    len(study_more_10) - len(overlap),
    len(attendance_more_80) - len(overlap),
    len(overlap)
),
set_labels = ('Study >10 Hours', 'Attendance >80'))

plt.title("Venn Diagram")
plt.show()
```

```python

# Question 5 - Contingency Table & Probability Calculations

contingency_table = pd.crosstab(
    student_data['group_discussion'],
    student_data['final_exam_pass']
)

print(contingency_table)

joint_probability = len(student_data[
    (student_data['group_discussion'] == 'Yes') &
    (student_data['final_exam_pass'] == 'Pass')
]) / len(student_data)

print("\nJoint Probability =", joint_probability)

marginal_probability = (
    student_data['final_exam_pass'] == 'Pass'
).mean()

print("Marginal Probability =", marginal_probability)

conditional_probability = len(student_data[
    (student_data['group_discussion'] == 'Yes') &
    (student_data['final_exam_pass'] == 'Pass')
]) / len(student_data[
    student_data['group_discussion'] == 'Yes'
])

print("Conditional Probability =", conditional_probability)
```

```python

# Question 6 - Understanding Relationships

print("Interpretation:")
print("Students participating in group discussions have a higher probability of passing.")

print("\nChecking Independence")

joint = joint_probability
independent_check = (
    ((student_data['group_discussion'] == 'Yes').mean()) *
    ((student_data['final_exam_pass'] == 'Pass').mean())
)

print("Joint Probability =", joint)
print("P(A)*P(B) =", independent_check)

if round(joint,2) == round(independent_check,2):
    print("Events are Independent")
else:
    print("Events are Dependent")
```

```python

# Question 7 - Bayes Theorem Application

# Given Data
P_H_given_P = 0.70
P_H_given_F = 0.40
P_H = 0.60

P_F = 1 - empirical_probability

# Bayes Theorem
# P(Pass | High Attendance)

numerator = P_H_given_P * empirical_probability

denominator = numerator + (P_H_given_F * P_F)

bayes_probability = numerator / denominator

print("Probability that student passed given high attendance =")
print(bayes_probability)
```
