# 1. Define the following with examples from the dataset

## Types of Data: Numerical & Categorical

### Numerical Data
Numerical data contains numbers and can be measured.

**Examples from dataset:**
- Age_of_Household_Head
- Household_Income
- Family_Size

### Categorical Data
Categorical data contains categories or labels.

**Examples from dataset:**
- Education_Level
- Owns_House
- Urban_Rural

---

## Types of Statistics: Descriptive vs Inferential

### Descriptive Statistics
Descriptive statistics summarize and describe the data.

**Examples:**
- Mean income
- Median age
- Histogram of income

### Inferential Statistics
Inferential statistics are used to make predictions or conclusions about a population based on sample data.

**Examples:**
- Predicting average income of a larger population
- Estimating family size trends

---

## What is Descriptive Statistics?

Descriptive statistics is the branch of statistics used to summarize, organize, and present data meaningfully.

It includes:
- Mean
- Median
- Mode
- Range
- Standard Deviation
- Charts and Graphs



# 2. Explain the Difference Between

## Mean, Median, and Mode

- **Mean:** Average of all values.
- **Median:** Middle value after sorting.
- **Mode:** Most frequently occurring value.

---

## Range, Variance, and Standard Deviation

- **Range:** Difference between maximum and minimum values.
- **Variance:** Average squared deviation from the mean.
- **Standard Deviation:** Square root of variance.



# 3. Important Statistical Terms

### Gaussian Distribution
A bell-shaped normal distribution curve.

### Log Normal Distribution
Distribution where logarithm of values follows normal distribution.

### 3-Sigma Rule
Almost 99.7% data lies within 3 standard deviations from the mean.

### Percentiles
Values below which a percentage of observations fall.

### Quartiles
Values dividing data into 4 equal parts.

### Five Number Summary
Minimum, Q1, Median, Q3, Maximum.

### Skewness
Measure of asymmetry in data distribution.

### Kurtosis
Measure of peakness in data distribution.



# Part B: Practical Answers

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('data/household_dataset.csv')

df.head()
```


# 3. Types of Data

## Point 1: Identify Numerical Columns
Numerical columns contain numbers and can perform mathematical operations.

## Point 2: Identify Categorical Columns
Categorical columns contain labels or categories.


```python

print("Data Types:\n")
print(df.dtypes)

numerical_cols = df.select_dtypes(include=np.number).columns
categorical_cols = df.select_dtypes(exclude=np.number).columns

print("\nNumerical Columns:")
print(list(numerical_cols))

print("\nCategorical Columns:")
print(list(categorical_cols))
```


# 4. Central Tendency

## Point 1: Calculate Mean, Median, and Mode
Mean, Median, and Mode are measures of central tendency.

## Point 2: Interpretation
These values help understand the center and distribution of household income and age.


```python

# Household Income
print("Household Income Statistics")
print("Mean:", df['Household_Income'].mean())
print("Median:", df['Household_Income'].median())
print("Mode:", df['Household_Income'].mode()[0])

print("\nAge Statistics")
print("Mean:", df['Age_of_Household_Head'].mean())
print("Median:", df['Age_of_Household_Head'].median())
print("Mode:", df['Age_of_Household_Head'].mode()[0])
```


# 5. Measures of Dispersion

## Point 1: Calculate Range, Variance, and Standard Deviation
These measures show how spread out the income data is.

## Point 2: Find IQR
IQR measures the spread of the middle 50% of data.


```python

income_range = df['Household_Income'].max() - df['Household_Income'].min()
income_variance = df['Household_Income'].var()
income_std = df['Household_Income'].std()

Q1 = df['Household_Income'].quantile(0.25)
Q3 = df['Household_Income'].quantile(0.75)
IQR = Q3 - Q1

print("Range:", income_range)
print("Variance:", income_variance)
print("Standard Deviation:", income_std)
print("IQR:", IQR)
```


# 6. Distribution

## Point 1: Plot Histogram of Household Income
Histogram shows the frequency distribution of income.

## Point 2: Calculate Skewness and Kurtosis
These values describe the shape of the income distribution.


```python

# Histogram
plt.hist(df['Household_Income'], bins=15)
plt.xlabel("Household Income")
plt.ylabel("Frequency")
plt.title("Histogram of Household Income")
plt.show()

# Skewness and Kurtosis
print("Skewness:", df['Household_Income'].skew())
print("Kurtosis:", df['Household_Income'].kurt())
```


# 7. Data Categorization

## Point 1: Compare Income by Education Level
Visualization helps compare income across education levels.

## Point 2: Compare Income by Urban and Rural Areas
Visualization shows income differences between Urban and Rural households.


```python

# Boxplot for Education Level
df.boxplot(column='Household_Income', by='Education_Level')
plt.title("Income by Education Level")
plt.suptitle("")
plt.xlabel("Education Level")
plt.ylabel("Income")
plt.show()
```

```python
# Histogram for Urban/Rural
df.hist(column='Household_Income', by='Urban_Rural')
plt.title("Income by Urban/Rural")
plt.suptitle("")
plt.xlabel("Area")
plt.ylabel("Income")
plt.show()
```
