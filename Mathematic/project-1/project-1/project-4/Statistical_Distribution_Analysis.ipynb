## 📚 Part A — Theoretical Foundation

### 1. What is Statistical Distributions?
A **statistical distribution** describes how values of a random variable are spread or distributed. It tells us the probability of each possible outcome. Distributions can be characterized by parameters such as mean, variance, and skewness. Examples include Normal, Binomial, Poisson, and Log-Normal distributions.

---

### 2. What is a Q-Q Plot and why is it used?
A **Q-Q (Quantile-Quantile) Plot** is a graphical tool to assess whether a dataset follows a particular theoretical distribution (usually Normal). It plots quantiles of the observed data against quantiles of the theoretical distribution.  
- **If the data follows the distribution**: points lie on a straight diagonal line.  
- **Deviations** from the line indicate skewness, heavy tails, or outliers.  
It is widely used to check normality before applying statistical tests.

---

### 3. Difference between Discrete and Continuous Distributions
| Feature | Discrete Distribution | Continuous Distribution |
|---|---|---|
| Values | Countable (0, 1, 2, ...) | Any real number in a range |
| Probability | P(X = x) > 0 possible | P(X = x) = 0; use intervals |
| Function | PMF (Probability Mass Function) | PDF (Probability Density Function) |
| Examples | Binomial, Poisson, Bernoulli | Normal, Log-Normal, Exponential |

---

### 4. What is Bernoulli Distribution?
The **Bernoulli Distribution** models a single experiment with exactly **two outcomes**: Success (1) with probability *p*, and Failure (0) with probability *1−p*.  
- **PMF**: P(X=1) = p, P(X=0) = 1−p  
- **Mean**: p | **Variance**: p(1−p)  
- **Example**: Whether a single transaction succeeds or fails.

---

### 5. What is Binomial Distribution?
The **Binomial Distribution** extends Bernoulli to *n* independent trials, counting the number of successes.  
- **PMF**: P(X=k) = C(n,k) · pᵏ · (1−p)ⁿ⁻ᵏ  
- **Mean**: np | **Variance**: np(1−p)  
- **Example**: Number of successful transactions in a week of 10 attempts.

---

### 6. Explain Log-Normal Distribution
A variable X follows a **Log-Normal Distribution** if ln(X) is normally distributed. It is used for data that is positively skewed and always positive.  
- **PDF**: Defined by μ (mean of log) and σ (std of log)  
- **Mean**: e^(μ + σ²/2) | **Variance**: (e^(σ²)−1)·e^(2μ+σ²)  
- **Example**: Transaction amounts — most are small but a few are very large.

---

### 7. Explain Power Law Distribution
A **Power Law Distribution** describes phenomena where large events are rare but possible, following P(X=x) ∝ x^(−α).  
- Characterized by a heavy tail — extreme values have non-negligible probability.  
- **Example**: Wealth distribution, internet traffic, and large transaction amounts.

---

### 8. What is Box-Cox Transform?
The **Box-Cox Transformation** is a family of power transformations used to **stabilize variance** and make data more normally distributed.  
- Formula: y(λ) = (xᵏ − 1)/λ for λ≠0; ln(x) for λ=0  
- The optimal λ is estimated by maximizing the log-likelihood.  
- **Use case**: Transforming skewed financial data before applying normal-assumption tests.

---

### 9. Explain Poisson Distribution with an example
The **Poisson Distribution** models the number of events occurring in a fixed interval of time/space, given a known average rate λ.  
- **PMF**: P(X=k) = e^(−λ) · λᵏ / k!  
- **Mean** = **Variance** = λ  
- **Example**: Number of transactions occurring in a single day. If the average is 7 transactions/day, Poisson(λ=7) models the daily count.

---

### 10. What is Z-score Probability?
A **Z-score** measures how many standard deviations a value is from the mean: Z = (X − μ) / σ.  
- Z-scores follow the Standard Normal Distribution N(0,1).  
- Used to compute probabilities like P(X > threshold) by looking up the standard normal CDF.  
- **Example**: Finding the probability that a transaction exceeds ₹5000.

---

### 11. Differentiate PDF and CDF
| Feature | PDF (Probability Density Function) | CDF (Cumulative Distribution Function) |
|---|---|---|
| Definition | Rate of probability at a point | Cumulative probability up to x |
| Formula | f(x) = dF(x)/dx | F(x) = P(X ≤ x) |
| Range | f(x) ≥ 0 | 0 ≤ F(x) ≤ 1 |
| Use | Find density at a value | Find probability up to a value |
| Integral | ∫f(x)dx = 1 | F(∞) = 1 |

## Task 1 — Bernoulli & Binomial Distributions
**Bernoulli**: Models whether a single transaction is successful (1) or failed (0).  
**Binomial**: Models the number of successful transactions in weekly counts.

```python
# ── Bernoulli Distribution ────────────────────────────────────────────────────
df['is_success'] = (df['transaction_status'] == 'Success').astype(int)
p_success = df['is_success'].mean()

print(f'=== BERNOULLI DISTRIBUTION ===')
print(f'p (success probability) = {p_success:.4f}')
print(f'q (failure probability) = {1 - p_success:.4f}')
print(f'Mean = {p_success:.4f}')
print(f'Variance = {p_success * (1 - p_success):.4f}')

bern_dist = bernoulli(p=p_success)

# ── Binomial Distribution ─────────────────────────────────────────────────────
# n = max weekly transaction count observed
n_max = int(df['transaction_count'].max())
binom_dist = binom(n=n_max, p=p_success)

print(f'\n=== BINOMIAL DISTRIBUTION ===')
print(f'n (trials) = {n_max}')
print(f'p (success prob) = {p_success:.4f}')
print(f'Mean = {binom_dist.mean():.4f}')
print(f'Variance = {binom_dist.var():.4f}')
print(f'Std Dev = {binom_dist.std():.4f}')
```

```python
# ── Plot Bernoulli & Binomial ─────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Bernoulli Bar Chart
outcomes = [0, 1]
probs = [1 - p_success, p_success]
axes[0].bar(['Fail (0)', 'Success (1)'], probs,
            color=['#e74c3c', '#2ecc71'], edgecolor='black', width=0.4)
axes[0].set_title(f'Bernoulli Distribution\np = {p_success:.4f}', fontsize=13, fontweight='bold')
axes[0].set_ylabel('Probability')
axes[0].set_ylim(0, 0.8)
for i, v in enumerate(probs):
    axes[0].text(i, v + 0.01, f'{v:.4f}', ha='center', fontsize=12, fontweight='bold')

# Interpretation
axes[0].text(0.5, -0.12,
    f'Interpretation: Each transaction has a {p_success*100:.1f}% chance of success.',
    ha='center', transform=axes[0].transAxes, fontsize=9, color='gray')

# Binomial PMF
k_vals = np.arange(0, n_max + 1)
pmf_vals = binom_dist.pmf(k_vals)
axes[1].bar(k_vals, pmf_vals, color='steelblue', edgecolor='black', alpha=0.8)
axes[1].set_title(f'Binomial Distribution (n={n_max}, p={p_success:.3f})',
                  fontsize=13, fontweight='bold')
axes[1].set_xlabel('Number of Successes')
axes[1].set_ylabel('Probability')

# Observed weekly counts overlay
obs_counts = df['transaction_count'].value_counts(normalize=True).sort_index()
axes[1].plot(obs_counts.index, obs_counts.values, 'ro-', markersize=6, label='Observed')
axes[1].legend()
axes[1].text(0.5, -0.12,
    f'Interpretation: Most weeks see {int(binom_dist.mean())} successes (mean={binom_dist.mean():.2f}).',
    ha='center', transform=axes[1].transAxes, fontsize=9, color='gray')

plt.suptitle('Task 1: Bernoulli & Binomial Distributions', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('bernoulli_binomial.png', bbox_inches='tight')
plt.show()
print('✅ Chart saved: bernoulli_binomial.png')
```

## Task 2 — Poisson Distribution
**Goal**: Fit the number of transactions per day to a Poisson distribution.  
Poisson is ideal when events occur independently at a constant average rate.

```python
# ── Poisson Distribution ──────────────────────────────────────────────────────
daily_counts = df.groupby('transaction_date').size()
lambda_est = daily_counts.mean()

print('=== POISSON DISTRIBUTION ===')
print(f'Number of days observed: {len(daily_counts)}')
print(f'Daily transaction counts:\n{daily_counts.describe()}')
print(f'\nEstimated λ (lambda) = {lambda_est:.4f}')
print(f'Mean = {lambda_est:.4f}')
print(f'Variance = {lambda_est:.4f}  (Mean = Variance is a Poisson property)')

poisson_dist = poisson(mu=lambda_est)
```

```python
# ── Plot Poisson ──────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# PMF of Poisson
k_range = np.arange(0, int(lambda_est * 2.5) + 1)
pmf_poisson = poisson_dist.pmf(k_range)

axes[0].bar(k_range, pmf_poisson, color='coral', edgecolor='black', alpha=0.8, label='Poisson PMF')

# Overlay observed
obs_daily = daily_counts.value_counts(normalize=True).sort_index()
axes[0].plot(obs_daily.index, obs_daily.values, 'b^-', markersize=7, label='Observed')
axes[0].axvline(lambda_est, color='darkred', linestyle='--', label=f'λ = {lambda_est:.2f}')
axes[0].set_title(f'Poisson Distribution (λ={lambda_est:.2f})', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Transactions per Day')
axes[0].set_ylabel('Probability')
axes[0].legend()
axes[0].text(0.5, -0.12,
    f'Interpretation: On average {lambda_est:.1f} transactions per day; Poisson fits daily counts well.',
    ha='center', transform=axes[0].transAxes, fontsize=9, color='gray')

# Daily count timeline
axes[1].bar(range(len(daily_counts)), daily_counts.values, color='teal', alpha=0.7)
axes[1].axhline(lambda_est, color='red', linestyle='--', linewidth=2, label=f'Mean = {lambda_est:.1f}')
axes[1].set_title('Daily Transaction Counts Over Time', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Day Index')
axes[1].set_ylabel('Transaction Count')
axes[1].legend()
axes[1].text(0.5, -0.12,
    'Interpretation: Transaction counts per day fluctuate around the mean, consistent with Poisson.',
    ha='center', transform=axes[1].transAxes, fontsize=9, color='gray')

plt.suptitle('Task 2: Poisson Distribution', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('poisson_distribution.png', bbox_inches='tight')
plt.show()
print('✅ Chart saved: poisson_distribution.png')
```

## Task 3 — Log-Normal & Power Law Distributions
**Goal**: Model transaction amounts using Log-Normal and Power Law distributions.  
Financial data is typically right-skewed — Log-Normal and Power Law are natural fits.

```python
# ── Log-Normal Fit ────────────────────────────────────────────────────────────
amounts = df['transaction_amount'].values

# Fit Log-Normal
sigma_ln, loc_ln, scale_ln = lognorm.fit(amounts, floc=0)
mu_ln = np.log(scale_ln)  # mean of log

print('=== LOG-NORMAL DISTRIBUTION ===')
print(f'μ (mean of log-amounts) = {mu_ln:.4f}')
print(f'σ (std  of log-amounts) = {sigma_ln:.4f}')
print(f'Fitted Mean  = {lognorm(sigma_ln, scale=scale_ln).mean():.2f}')
print(f'Fitted Std   = {lognorm(sigma_ln, scale=scale_ln).std():.2f}')
print(f'Actual Mean  = {amounts.mean():.2f}')
print(f'Actual Std   = {amounts.std():.2f}')

# Fit Power Law
alpha_pl, loc_pl, scale_pl = powerlaw.fit(amounts, floc=0)
print(f'\n=== POWER LAW DISTRIBUTION ===')
print(f'Alpha (shape) = {alpha_pl:.4f}')
print(f'Scale         = {scale_pl:.4f}')
```

```python
# ── Plot Log-Normal & Power Law ───────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

x = np.linspace(amounts.min(), amounts.max(), 500)

# Log-Normal
axes[0].hist(amounts, bins=40, density=True, alpha=0.6, color='skyblue',
             edgecolor='white', label='Observed')
pdf_ln = lognorm(sigma_ln, scale=scale_ln).pdf(x)
axes[0].plot(x, pdf_ln, 'r-', linewidth=2.5, label=f'Log-Normal (μ={mu_ln:.2f}, σ={sigma_ln:.2f})')
axes[0].set_title('Log-Normal Fit — Transaction Amounts', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Transaction Amount (₹)')
axes[0].set_ylabel('Density')
axes[0].legend()
axes[0].text(0.5, -0.12,
    f'Interpretation: Transaction amounts are right-skewed; Log-Normal captures the long tail well.',
    ha='center', transform=axes[0].transAxes, fontsize=9, color='gray')

# Power Law
axes[1].hist(amounts, bins=40, density=True, alpha=0.6, color='lightsalmon',
             edgecolor='white', label='Observed')
pdf_pl = powerlaw(alpha_pl, scale=scale_pl).pdf(x)
axes[1].plot(x, pdf_pl, 'b-', linewidth=2.5, label=f'Power Law (α={alpha_pl:.2f})')
axes[1].set_title('Power Law Fit — Transaction Amounts', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Transaction Amount (₹)')
axes[1].set_ylabel('Density')
axes[1].legend()
axes[1].text(0.5, -0.12,
    'Interpretation: Power Law shows heavy-tail behaviour — large transactions are more common than normal.',
    ha='center', transform=axes[1].transAxes, fontsize=9, color='gray')

plt.suptitle('Task 3: Log-Normal & Power Law Distributions', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('lognormal_powerlaw.png', bbox_inches='tight')
plt.show()
print('✅ Chart saved: lognormal_powerlaw.png')
```

## Task 4 — Q-Q Plot (Normality Test)
**Goal**: Generate a Q-Q plot to test whether transaction amounts are normally distributed.

```python
# ── Q-Q Plot ──────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Q-Q for raw amounts
ax = axes[0]
(osm, osr), (slope, intercept, r) = probplot(amounts, dist='norm')
ax.plot(osm, osr, 'o', color='steelblue', markersize=4, alpha=0.7, label='Data Quantiles')
ax.plot(osm, slope * np.array(osm) + intercept, 'r-', linewidth=2, label='Normal Reference Line')
ax.set_title('Q-Q Plot: Raw Transaction Amounts', fontsize=13, fontweight='bold')
ax.set_xlabel('Theoretical Quantiles')
ax.set_ylabel('Sample Quantiles')
ax.legend()
ax.text(0.5, -0.15,
    f'R² = {r**2:.4f} | Interpretation: Points deviate from line → data is NOT normally distributed (right skew).',
    ha='center', transform=ax.transAxes, fontsize=9, color='gray')

# Q-Q for log-transformed amounts
log_amounts = np.log(amounts)
ax2 = axes[1]
(osm2, osr2), (slope2, intercept2, r2) = probplot(log_amounts, dist='norm')
ax2.plot(osm2, osr2, 'o', color='mediumseagreen', markersize=4, alpha=0.7, label='Log-Data Quantiles')
ax2.plot(osm2, slope2 * np.array(osm2) + intercept2, 'r-', linewidth=2, label='Normal Reference Line')
ax2.set_title('Q-Q Plot: Log-Transformed Amounts', fontsize=13, fontweight='bold')
ax2.set_xlabel('Theoretical Quantiles')
ax2.set_ylabel('Sample Quantiles')
ax2.legend()
ax2.text(0.5, -0.15,
    f'R² = {r2**2:.4f} | Interpretation: Log-transformed amounts closely follow normality (points near line).',
    ha='center', transform=ax2.transAxes, fontsize=9, color='gray')

# Shapiro-Wilk Test
stat, p_val = stats.shapiro(amounts[:50])  # shapiro works best on n<50
print(f'Shapiro-Wilk Test (raw amounts): W={stat:.4f}, p={p_val:.4f}')
print('→ p < 0.05: Data is NOT normally distributed.' if p_val < 0.05 else '→ p ≥ 0.05: Data appears normal.')

plt.suptitle('Task 4: Q-Q Plot — Normality Test', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('qqplot.png', bbox_inches='tight')
plt.show()
print('✅ Chart saved: qqplot.png')
```

## Task 5 — Box-Cox Transformation
**Goal**: Apply Box-Cox Transform to stabilize variance in transaction amounts.

```python
# ── Box-Cox Transform ─────────────────────────────────────────────────────────
transformed, lambda_bc = boxcox(amounts)

print('=== BOX-COX TRANSFORMATION ===')
print(f'Optimal Lambda (λ) = {lambda_bc:.4f}')
print(f'\nBefore Transform:')
print(f'  Mean     = {amounts.mean():.4f}')
print(f'  Variance = {amounts.var():.4f}')
print(f'  Skewness = {stats.skew(amounts):.4f}')
print(f'\nAfter Box-Cox:')
print(f'  Mean     = {transformed.mean():.4f}')
print(f'  Variance = {transformed.var():.4f}')
print(f'  Skewness = {stats.skew(transformed):.4f}')
```

```python
# ── Plot Box-Cox ──────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].hist(amounts, bins=40, color='steelblue', edgecolor='white', alpha=0.8)
axes[0].set_title('Before Box-Cox: Raw Transaction Amounts', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Transaction Amount (₹)')
axes[0].set_ylabel('Frequency')
axes[0].text(0.05, 0.95,
    f'Skewness: {stats.skew(amounts):.3f}',
    transform=axes[0].transAxes, fontsize=10, va='top',
    bbox=dict(boxstyle='round', fc='lightyellow', ec='orange'))
axes[0].text(0.5, -0.12,
    'Interpretation: Highly right-skewed distribution before transformation.',
    ha='center', transform=axes[0].transAxes, fontsize=9, color='gray')

axes[1].hist(transformed, bins=40, color='coral', edgecolor='white', alpha=0.8)
axes[1].set_title(f'After Box-Cox (λ={lambda_bc:.3f}): Transformed Amounts',
                  fontsize=13, fontweight='bold')
axes[1].set_xlabel('Transformed Value')
axes[1].set_ylabel('Frequency')
axes[1].text(0.05, 0.95,
    f'Skewness: {stats.skew(transformed):.3f}',
    transform=axes[1].transAxes, fontsize=10, va='top',
    bbox=dict(boxstyle='round', fc='lightgreen', ec='green'))
axes[1].text(0.5, -0.12,
    'Interpretation: Skewness is significantly reduced; distribution is closer to normal.',
    ha='center', transform=axes[1].transAxes, fontsize=9, color='gray')

plt.suptitle('Task 5: Box-Cox Transformation', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('boxcox_transform.png', bbox_inches='tight')
plt.show()
print('✅ Chart saved: boxcox_transform.png')
```

## Task 6 — Z-Score Analysis
**Goal**: Calculate Z-scores for transaction amounts and compute the probability of transactions exceeding ₹5000.

```python
# ── Z-Score Analysis ──────────────────────────────────────────────────────────
mu_amt = amounts.mean()
sigma_amt = amounts.std()

df['z_score'] = (df['transaction_amount'] - mu_amt) / sigma_amt

threshold = 5000
z_threshold = (threshold - mu_amt) / sigma_amt
prob_above = 1 - norm.cdf(z_threshold)
observed_prob = (amounts > threshold).mean()

print('=== Z-SCORE ANALYSIS ===')
print(f'Mean Amount (μ)  = ₹{mu_amt:.2f}')
print(f'Std Dev   (σ)    = ₹{sigma_amt:.2f}')
print(f'\nZ-score for ₹{threshold}: {z_threshold:.4f}')
print(f'Theoretical P(X > ₹{threshold}) = {prob_above:.4f} ({prob_above*100:.2f}%)')
print(f'Observed   P(X > ₹{threshold}) = {observed_prob:.4f} ({observed_prob*100:.2f}%)')

print('\nTop 5 highest Z-score transactions:')
print(df[['transaction_id', 'transaction_amount', 'z_score']]
      .nlargest(5, 'z_score')[['transaction_amount', 'z_score']].to_string())
```

```python
# ── Plot Z-Score ──────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Z-score distribution
axes[0].hist(df['z_score'], bins=40, color='mediumslateblue',
             edgecolor='white', alpha=0.8, density=True)
x_z = np.linspace(df['z_score'].min(), df['z_score'].max(), 300)
axes[0].plot(x_z, norm.pdf(x_z), 'r-', linewidth=2, label='Standard Normal N(0,1)')
axes[0].axvline(z_threshold, color='darkred', linestyle='--', linewidth=2,
                label=f'Z(₹5000) = {z_threshold:.2f}')
axes[0].fill_betweenx([0, norm.pdf(z_threshold)],
                       z_threshold, df['z_score'].max(),
                       alpha=0.2, color='red', label=f'P > ₹5000 = {prob_above:.2%}')
axes[0].set_title('Z-Score Distribution', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Z-Score')
axes[0].set_ylabel('Density')
axes[0].legend(fontsize=8)
axes[0].text(0.5, -0.12,
    f'Interpretation: {prob_above*100:.1f}% of transactions exceed ₹5000 (Z>{z_threshold:.2f}).',
    ha='center', transform=axes[0].transAxes, fontsize=9, color='gray')

# Scatter: amount vs z-score
colors = ['green' if x <= threshold else 'red' for x in amounts]
axes[1].scatter(df['transaction_amount'], df['z_score'], c=colors, alpha=0.5, s=20)
axes[1].axhline(z_threshold, color='darkred', linestyle='--', label=f'Z = {z_threshold:.2f} (₹5000)')
axes[1].axvline(threshold, color='orange', linestyle='--', label='₹5000 threshold')
axes[1].set_title('Transaction Amount vs Z-Score', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Transaction Amount (₹)')
axes[1].set_ylabel('Z-Score')
axes[1].legend(fontsize=8)
axes[1].text(0.5, -0.12,
    'Red dots = transactions above ₹5000 (outliers); Green = below threshold.',
    ha='center', transform=axes[1].transAxes, fontsize=9, color='gray')

plt.suptitle('Task 6: Z-Score Probability Analysis', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('zscore_analysis.png', bbox_inches='tight')
plt.show()
print('✅ Chart saved: zscore_analysis.png')
```

## Task 7 — PDF and CDF of Transaction Amounts
**Goal**: Plot and interpret the Probability Density Function and Cumulative Distribution Function.

```python
# ── PDF & CDF ─────────────────────────────────────────────────────────────────
x_range = np.linspace(amounts.min(), amounts.max(), 500)

# KDE for empirical PDF
kde = stats.gaussian_kde(amounts)
pdf_empirical = kde(x_range)

# Fitted Log-Normal PDF & CDF
ln_dist = lognorm(sigma_ln, scale=scale_ln)
pdf_lognorm = ln_dist.pdf(x_range)
cdf_lognorm = ln_dist.cdf(x_range)

# Empirical CDF
sorted_amounts = np.sort(amounts)
ecdf = np.arange(1, len(sorted_amounts) + 1) / len(sorted_amounts)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# PDF
axes[0].hist(amounts, bins=40, density=True, alpha=0.4, color='skyblue',
             edgecolor='white', label='Histogram')
axes[0].plot(x_range, pdf_empirical, 'b-', linewidth=2.5, label='KDE (Empirical PDF)')
axes[0].plot(x_range, pdf_lognorm, 'r--', linewidth=2, label='Log-Normal PDF')
axes[0].axvline(mu_amt, color='darkgreen', linestyle='-.',
                label=f'Mean = ₹{mu_amt:.0f}')
axes[0].set_title('PDF — Transaction Amounts', fontsize=13, fontweight='bold')
axes[0].set_xlabel('Transaction Amount (₹)')
axes[0].set_ylabel('Density')
axes[0].legend(fontsize=8)
axes[0].text(0.5, -0.12,
    f'Interpretation: PDF peaks around ₹{amounts[np.argmax(pdf_empirical):.0f]} and shows positive skew.',
    ha='center', transform=axes[0].transAxes, fontsize=9, color='gray')

# CDF
axes[1].plot(sorted_amounts, ecdf, 'b-', linewidth=2.5, label='Empirical CDF (ECDF)')
axes[1].plot(x_range, cdf_lognorm, 'r--', linewidth=2, label='Log-Normal CDF')
axes[1].axvline(5000, color='darkred', linestyle='--', label='₹5000 threshold')
p_5000 = ln_dist.cdf(5000)
axes[1].axhline(p_5000, color='orange', linestyle=':', label=f'P(X≤5000) = {p_5000:.2%}')
axes[1].set_title('CDF — Transaction Amounts', fontsize=13, fontweight='bold')
axes[1].set_xlabel('Transaction Amount (₹)')
axes[1].set_ylabel('Cumulative Probability')
axes[1].legend(fontsize=8)
axes[1].text(0.5, -0.12,
    f'Interpretation: {p_5000*100:.1f}% of transactions are ≤₹5000; {(1-p_5000)*100:.1f}% exceed ₹5000.',
    ha='center', transform=axes[1].transAxes, fontsize=9, color='gray')

plt.suptitle('Task 7: PDF & CDF of Transaction Amounts', fontsize=15, fontweight='bold', y=1.02)
plt.tight_layout()
plt.savefig('pdf_cdf.png', bbox_inches='tight')
plt.show()
print('✅ Chart saved: pdf_cdf.png')
```

## Task 8 — Best Distribution Fit & Conclusion

```python
# ── Goodness-of-Fit: AIC / Log-Likelihood ────────────────────────────────────
from scipy.stats import kstest

distributions = {
    'Normal':     norm(*norm.fit(amounts)),
    'Log-Normal': lognorm(*lognorm.fit(amounts, floc=0)),
    'Power Law':  powerlaw(*powerlaw.fit(amounts, floc=0)),
}

print('=== GOODNESS-OF-FIT COMPARISON (Kolmogorov-Smirnov Test) ===')
print(f'{"Distribution":<15} {"KS Statistic":<15} {"p-value":<15} {"Verdict"}')
print('-' * 60)
results = {}
for name, dist in distributions.items():
    ks_stat, p_val = kstest(amounts, dist.cdf)
    verdict = 'GOOD FIT ✅' if p_val > 0.05 else 'Poor Fit ❌'
    results[name] = (ks_stat, p_val)
    print(f'{name:<15} {ks_stat:<15.4f} {p_val:<15.4f} {verdict}')
```

```python
# ── Final Summary Plot ────────────────────────────────────────────────────────
fig, ax = plt.subplots(figsize=(12, 6))

ax.hist(amounts, bins=50, density=True, alpha=0.4, color='lightblue',
        edgecolor='white', label='Observed Amounts')

colors_map = {'Normal': 'red', 'Log-Normal': 'green', 'Power Law': 'purple'}
for name, dist in distributions.items():
    ax.plot(x_range, dist.pdf(x_range),
            color=colors_map[name], linewidth=2.5, label=f'{name} (KS={results[name][0]:.4f})')

ax.set_title('Task 8: Distribution Comparison — Best Fit for Transaction Amounts',
             fontsize=13, fontweight='bold')
ax.set_xlabel('Transaction Amount (₹)')
ax.set_ylabel('Density')
ax.legend(fontsize=10)

best = min(results, key=lambda k: results[k][0])
ax.text(0.5, -0.12,
    f'✅ Best Fit: {best} (lowest KS statistic = {results[best][0]:.4f}) — confirmed by Q-Q plot and visual inspection.',
    ha='center', transform=ax.transAxes, fontsize=10, color='darkgreen', fontweight='bold')

plt.tight_layout()
plt.savefig('best_fit_comparison.png', bbox_inches='tight')
plt.show()
print('✅ Chart saved: best_fit_comparison.png')
```
