## Types of Data: Numerical & Categorical

### Numerical Data
Numerical data contains numbers and can be measured.

**Examples from dataset:**
- Age_of_Household_Head
- Household_Income
- Family_Size

### Categorical Data
Categorical data contains categories or labels.

**Examples from dataset:**
- Education_Level
- Owns_House
- Urban_Rural

---

## Types of Statistics: Descriptive vs Inferential

### Descriptive Statistics
Descriptive statistics summarize and describe the data.

**Examples:**
- Mean income
- Median age
- Histogram of income

### Inferential Statistics
Inferential statistics are used to make predictions or conclusions about a population based on sample data.

**Examples:**
- Predicting average income of a larger population
- Estimating family size trends

---

## What is Descriptive Statistics?

Descriptive statistics is the branch of statistics used to summarize, organize, and present data meaningfully.

It includes:
- Mean
- Median
- Mode
- Range
- Standard Deviation
- Charts and Graphs



# 2. Explain the Difference Between

## Mean, Median, and Mode

- **Mean:** Average of all values.
- **Median:** Middle value after sorting.
- **Mode:** Most frequently occurring value.

---

## Range, Variance, and Standard Deviation

- **Range:** Difference between maximum and minimum values.
- **Variance:** Average squared deviation from the mean.
- **Standard Deviation:** Square root of variance.



# 3. Important Statistical Terms

### Gaussian Distribution
A bell-shaped normal distribution curve.

### Log Normal Distribution
Distribution where logarithm of values follows normal distribution.

### 3-Sigma Rule
Almost 99.7% data lies within 3 standard deviations from the mean.

### Percentiles
Values below which a percentage of observations fall.

### Quartiles
Values dividing data into 4 equal parts.

### Five Number Summary
Minimum, Q1, Median, Q3, Maximum.

### Skewness
Measure of asymmetry in data distribution.

### Kurtosis
Measure of peakness in data distribution.
